# Structure de données : le dictionnaire python

La table de correspondance, ou dictionnaire en python, est une structure de données semblable aux listes.

Il y a deux différences principales par rapport aux listes :
- les indices des éléments (appelés clés) peuvent être de types bien plus variés, pas uniquement entier (ex: string)
- les éléments ne sont pas ordonnés

## Créer un dictionaire

### Un dictionnaire vide

````python
my_dictionary = {}
print(my_dictionary)
print(type(my_dictionary))
````

### Peupler un dictionnaire

````python
my_dictionary["indice"] = "valeur" # on ajoute un élément à un dictionnaire
print(my_dictionary["indice"]) # à présent, "indice" se comporte comme un indice dans une liste, on peut y accéder avec la même syntaxe

# pareil que précédemment, mais avec un nombre
my_dictionary[42] = "La grande question sur la vie, l'univers et le reste"
print(my_dictionary[42])

my_dictionary["La grande question sur la vie, l'univers et le reste"] = 42
print(my_dictionary["La grande question sur la vie, l'univers et le reste"])
print(my_dictionary)
````

````python
# on peut aussi créer un dictionnaire avec des éléments déjà présent
# on peut écrire sur plusieurs lignes pour plus de clarté
avoir = {
    "je": "ai",
    "j'": "ai",
    "tu": "as",
    "il": "a",
    "elle": "a",
    "nous": "avons",
    "vous": "avez",
    "ils": "ont",
    "elles": "ont"
}
print (avoir["tu"])
print(avoir)
#ajouter le verbe être
````

### Compréhension de dictionnaire

On peut également créer un dictionnaire à l'aide d'une compréhension de dictionnaire. Cette construction est analogue à la compréhension de liste.

Les deux méthodes présentées donnent ici le même résultat, seule la façon de voir les éléments dans la compréhension change.

````python
avoir_liste = [
    ["je", "ai"],
    ["j'", "ai"],
    ["tu", "as"],
    ["il", "a"],
    ["elle", "a"],
    ["nous", "avons"],
    ["vous", "avez"],
    ["ils", "ont"],
    ["elles", "ont"]
]
print (avoir_liste)
print (avoir["tu"])
print (len(avoir_liste))
````

````python
# on crée l'ensemble des clés-valeur en récupérant les éléments intéressants via leur index
avoir_dictionnaire1 = {element[0]: element[0] for element in avoir_liste}
print (len(avoir_dictionnaire1))
print(avoir_dictionnaire1)
````

````python
# on crée l'ensemble des clés-valeur en dépliant les éléments d'une liste
avoir_dictionnaire2 = {personne: flexion for personne, flexion in avoir_liste}

print(avoir_dictionnaire2)
````

### Utiliser un dictionnaire

````python
print(["j'", avoir["j'"], "des", "pommes"])
print(["tu", avoir["tu"], "des", "pommes"])
````

Comme pour les listes, on peut faire des compréhensions pour créer des dictionnaires à partir d'un ensemble de valeurs

````python
liste = [["je", "ai"], ["j'", "ai"], ["tu", "as"], ["il", "a"], ["elle", "a"], ["nous", "avons"], ["vous", "avez"], ["ils", "ont"], ["elles", "ont"]]
avoir = {element[0]: element[1] for element in liste}
print(avoir)
print(["j'", avoir["j'"], "des", "pommes"])
print(["tu", avoir["tu"], "des", "pommes"])
print(["ils", avoir["eux"], "des", "pommes"])
````

Si on essaie d'accéder à une valeur qui n'existe pas, python renvoit une erreur

````python
combien_pommes = {"Alice" : 5, "Bob" : 3}
````

### Accéder à un élément absent

Lorsque l'on souhaite accéder à un élément absent du dictionnaire, python renvoie une erreur.

Ce comportement est **préférable** car cela indique très souvent une erreur dans votre programme.

````python
print(combien_pommes["Catherine"])
````

### Quand une erreur n'est pas souhaitable

Parfois, une erreur n'est pas souhaitable, une valeur **logique dans le cadre du programme** peut être utilisée si une clé n'existe pas.

Par exemple, si une personne n'est pas dans le dictionnaire "combien_pommes", on peut supposer que cette personne a 0 pommes.

Pour cela, deux solutions :
1. vérifier que la clé existe avant d'y accéder avec `in` :
```python
if cle in dictionnaire:
       # tout va bien
else:
       # rien ne va plus```

2. on utilise la méthode [get](https://docs.python.org/fr/3/library/stdtypes.html#dict.get)

#### Exercice

````python
qui = "Alice"
print("{} a {} pomme(s).".format(qui, combien_pommes.get(qui, 0)))
````

### Les méthodes utiles pour les dictionnaires

Pour accéder à l'ensemble des clés d'un dictionnaire, on utilise la méthode [keys](https://docs.python.org/fr/3/library/stdtypes.html#dict.keys)

````python
print(combien_pommes.keys())
````

````python
for key in combien_pommes.keys():
    print(key)
````

De base, itérer sur un dictionnaire signifie itérer sur ses clés.

````python
for key in combien_pommes:
    print(key)
````

Pour accéder à l'ensemble des valeurs d'un dictionnaire, on utilise la méthode [values](https://docs.python.org/fr/3/library/stdtypes.html#dict.values)

````python
print(combien_pommes.values())
````

````python
for value in combien_pommes.values():
    print(value)
````

Pour accéder à l'ensemble des couples clé/valeur d'un dictionnaire, on utilise la méthode [items](https://docs.python.org/fr/3/library/stdtypes.html#dict.items)

````python
print(combien_pommes.items())
````

````python
for element in combien_pommes.items():
    print(element)
````

````python
for clef, valeur in combien_pommes.items():
    print(clef, "->", valeur)
````
