```python
#exo1 commentez le code qui construit la loi de Zipf à partir du texte ci-dessous
import re
import matplotlib.pyplot as plt
```

```python
texte = "et la fée l'appela en lui disant : 'je vous donne un don'. à chaque parole que vous direz, il vous sortira de la bouche ou une fleur ou une pierre précieuse. Lorsque cette fille arriva au logis. Sa mère la gronda de revenir si tard de la fontaine 'je vous demande pardon ma mère d'avoir tardé si longtemps' dit cette fille. Et en disant ces mots il sortit de sa bouche deux roses deux perles et deux gros diamants."
```

```python
texte = texte.lower()
print(texte)
```

```python
liste_mots = re.split(', | |\. ', texte)
print(liste_mots)
```

```python
dictionnaire = {}
```

```python
#Find the frequency of words
for txt in liste_mots:
  if txt in dictionnaire.keys():
    dictionnaire[txt]+=1
  else:
    dictionnaire[txt]=1
```

```python
print (dictionnaire)
```

```python
frequences_mots={}
```

```python
frequences_mots = dict( #Sort the word frequencies in descending order
        sorted(
            dictionnaire.items(),
            key=lambda x: x[1],
            reverse=True)
        )
```

```python
print(frequences_mots)
```

```python
#Define two lists, rank and frequency
rang = []
frequence = []
init = 0
```

```python
#Assign ranks based on frequencies of words
for freq in frequences_mots.values():
  init+=1
  rang.append(init)
  frequence.append(freq)
```

```python
plt.plot(rang,frequence)
```

```python

plt.xlabel('Rang(r)')
plt.ylabel('Fréquence(f)')


plt.title("Loi de Zipf")
plt.plot(rang,frequence, color='red', linewidth=5, linestyle='dotted')
plt.show()
```

```python
plt.xscale('log')
plt.plot(rang,frequence)
plt.xlabel('Rang(r)')
plt.ylabel('Fréquence(f)')


plt.title("Loi de Zipf")
plt.show()
#fin exo1
```
