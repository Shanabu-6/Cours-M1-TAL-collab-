# Les objets

Les objets en python permettent de créer de nouveaux types.

Ces types ont deux utilités pour nous :
- créer de nouvelles fonctionalités spécifiques (ex: listes, dictionnaires)
- organiser des valeurs de manière logique (champs) qui autrement seraient dans des listes/tuples.

## Déclarer une classe vide

```python
# le mot-clé class permet de déclarer une classe
# comme def permet de déclarer une fonction
# le nom de la classe commence par une majuscule
class A:
    pass # veut dire qu'il n'y a rien dans la classe

# une fois qu'on a une classe, on peut créer des instances (variables)
# comme on le fait avec les int, str, list, dict, etc...
# on déclare une nouvelle instance en appelant la classe comme une fonction
element = A()

print(element)
```

# rajouter des champs à une classe

La fonction `__init__` permet d'initialiser les champs nécessaires à une classe.

Une méthode dans une classe commence toujours par `self` (soi). `self` est l'objet en cours. Cela permet la notation préfixée `objet.méthode`.

Quand vous avez une liste `l` et que vous faites `l.append(1)`, il existe une fonction `append` dans la classe List où le premier argument est `self`.

```python
class B:
    def __init__(self, truc):
        # de manière générale, on essaye de donner le même nom
        # à la variable qu'au champs en question, mais pas toujours !
        self.valeur1 = truc

element = B(2)
print(element)
print(element.valeur1)
```

# Exemple vu et revu : le rectangle

Un rectangle est une figure géométrique simple.

Un rectangle a quatre côtés. Comme les deux côtés opposés on la même longueur, on donnera deux valeurs : sa longueur et sa largeur.

On peut calculer pour un rectangle :
- son périmètre
- sa surface
- s'il est un carré

```python
class Rectangle:
    def __init__(self, longueur, largeur):
        self.longueur = longueur
        self.largeur = largeur

    def perimetre(self):
        return 2*(self.longueur) + 2*(self.largeur)

    def surface(self):
        return self.longueur * self.largeur

    def is_carre(self):
        return self.longueur == self.largeur

mon_rectangle = Rectangle(1, 2)
mon_carre = Rectangle(1, 1)

print(mon_rectangle.perimetre()) # 6
print(mon_rectangle.surface()) # 2
print(mon_rectangle.is_carre()) # False
print()
print(mon_carre.perimetre()) # 4
print(mon_carre.surface()) # 1
print(mon_carre.is_carre()) # True
```
