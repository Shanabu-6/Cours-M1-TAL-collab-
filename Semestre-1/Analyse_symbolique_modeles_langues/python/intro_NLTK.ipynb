```python
etiquetage_POS = nltk.pos_tag(tokens)
etiquetage_POS[0:10]
```

```python
import nltk
from nltk.collocations import *
bigram_measures = nltk.collocations.BigramAssocMeasures()
trigram_measures = nltk.collocations.TrigramAssocMeasures()
fourgram_measures = nltk.collocations.QuadgramAssocMeasures()
print (bigram_measures)
print (fourgram_measures)
```

```python
finder = BigramCollocationFinder.from_words(gutenberg.words('melville-moby_dick.txt'))
finder.nbest(bigram_measures.pmi, 10)
```

```python
finder = TrigramCollocationFinder.from_words(gutenberg.words('melville-moby_dick.txt'))
finder.nbest(trigram_measures.pmi, 10)
```
