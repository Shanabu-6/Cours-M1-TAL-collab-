```python
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Textes
A = "Les feuilles tombent à l'automne. Les feuilles sont jaunes et rouges. Les feuilles forment un tapis rouge et jaune"

B = "Les feuilles et les fleurs poussent au printemps. Les feuilles sont vertes. Les feuilles ne forment plus un tapis rouge et jaune"

corpus = [A, B]
```

```python

# sac de mots (bag of words en anglais)
sac = CountVectorizer(lowercase=True) 
sac_2 = sac.fit_transform(corpus).toarray() #créer une matrice ayant comme poids la fréq d'occurrence des mots

print("Vocabulaire sans ordre syntaxique =\n", sac.get_feature_names_out()) # pour aller plus loin : https://scikit-learn.org/stable/modules/generated/sklearn.preprocessing.OneHotEncoder.html
print("vectorisation sac de mots:\n", sac_2)

```

```python
# calcul similarité Jaccard -----
def jaccard(a, b):
    set_a = set(a.split())
    set_b = set(b.split())
    return len(set_a & set_b) / len(set_a | set_b)

print("Jaccard similarity =", jaccard(A.lower(), B.lower()))

```
